```python
import numpy as np
import pandas as pd

print("Numpy version :", np.__version__)
print("Pandas version :", pd.__version__)
```

```python
marks=np.array([10,20,30,40,50])
print("Marks",marks)
print("Data Type", type(marks))
```

```python
print("Dimension",marks.ndim)
print("Shape",marks.shape)
print("Size",marks.size)
```

```python
print("Marks : ", marks)
print("Add 5 : ", marks+5)
print("Multiply 2 : ", marks*2)
print("Divide by 2 : ", marks/2)
print("Square : ", marks ** 2)
```

```python
print("Sum : ",  marks.sum())
print("Mean : ",  marks.mean())
print("Max : ",  marks.max())
print("Min : ",  marks.min())
```

```python
print("First mark : ", marks[0])
print("Last mark : ", marks[-1])
```

```python
print("Index 1 to 3 : ", marks[1:4])
print("First 3 : ", marks[:3])
print("Index 2 to end : ", marks[2:])
print("Every 2nd number", marks[::2])
```

```python
student_marks=np.array([
    [78,85,92],
    [67,88,76],
    [90,95,89]
])
print(student_marks)
print("Shape :", student_marks.shape)
```

```python
print("Mean of all values : ", student_marks.mean())
print("Mean of each column : ", student_marks.mean(axis=0))
print("Mean of each row : ", student_marks.mean(axis=1))
```

```python
data={
    "Name":["Alice","Bob","Charlie","Sam"],
    "Branch":["CSE","ISE","AIML","CSE-AIML"],
    "Marks":[88,97,65,44]
}

df = pd.DataFrame(data)

print(df)
```

```python
print(df.shape)
print("Col names : ", df.columns)
```

```python
df.head() # top 5

```

```python
df.describe() #statistics
```

```python
print(df["Marks"]) #only marks
print()
print(df[["Marks","Branch"]])
```

ILOC

```python
# 1st row iloc

print(df.iloc[0])
print()
# 2nd row
print(df.iloc[1])
print()
#First three rows
print(df.iloc[0:3]) # 0,1,2
```

DF slicing

```python
# first 2 rows and first two cols
print(df.iloc[0:2,0:2])
```

```python
#             row, col
print(df.iloc[1:4, 0:3])
```

```python
print(df.iloc[0:2,0:2],df.iloc[3:,0:2])
```

FILTERING

```python
CSE=df[df["Branch"]=="CSE"]
print(CSE)
print()
print("90 nd above")
high_scorers=df[df["Marks"]>90]
print(high_scorers)
```
